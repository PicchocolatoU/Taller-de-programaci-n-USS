Desarrolla un script que pida un número al usuario e identifique si el valor ingresado es
positivo, negativo o cero.

```python
def verificar_numero():
    try:
        numero = float(input("Introduce un número: "))
        if numero > 0:
            print("El valor ingresado es positivo.")
        elif numero < 0:
            print("El valor ingresado es negativo.")
        else:
            print("El valor ingresado es cero.")

    except ValueError:
        print("Introduzca un númeron no otra cosa.")
        verificar_numero()

verificar_numero()
```

peso = masa × gravedad

Usa los siguientes valores:

• Tierra: 9.8 m/s²

• Luna: 1.62 m/s²

• Marte: 3.71 m/s²

• Júpiter: 24.79 m/s²

El programa debe mostrar el peso calculado para cada objeto.

Al final, debe mostrar:

• Cantidad de veces que se eligió cada cuerpo celeste

```python
def bolsa():
    cant_tierra = 0
    cant_luna = 0
    cant_marte = 0
    cant_jupiter = 0
    suma_pesos = 0

    while True:
        try:
            cant_analizar = int(input("Ingrese la cantidad de objetos que desea analizar"))
            if cant_analizar > 0:
                break
            else:
                print("ERROR: Ingrese una cantidad positiva")
        except ValueError:
                print("ERROR: Ingrese un número entero")

    for i in range(cant_analizar):
         print(f"CUERPO CELESTE {i + 1}: ")

         while True:          
                   cuerpo_celeste = input("Ingrese el cuerpo celeste que desea analizar (tierra, luna, marte o jupiter)").strip().lower()
                   if cuerpo_celeste in ["tierra", "luna", "marte", "jupiter"]:
                      print(f"cuerpo celeste escogido: {cuerpo_celeste}")
                      break
                   else:
                        print("ERRROR: Ingrese un cuerpo celeste válido para continuar")   

         while True:
                try:
                   kg = float(input("Ingrese la masa del cuerpo celeste en kg"))
                   if kg > 0:
                      print(f"masa ingresada: {kg} kg")
                      break
                   else:
                        print("ERRROR: Ingrese una cantidad positiva mayor a 0 para continuar")
                except ValueError:
                        print("ERROR: Ingrese un número entero positivo para continuar")

         


         if cuerpo_celeste == "tierra":
              peso_tierra = kg * 9.8
              print(f"el peso del cuerpo celeste elegido es: {peso_tierra}")
              cant_tierra = cant_tierra + 1
              suma_pesos = suma_pesos + peso_tierra

         elif cuerpo_celeste == "luna":
              peso_luna = kg * 1.62
              print(f"el peso del cuerpo celeste elegido es: {peso_luna}")
              cant_luna = cant_luna + 1
              suma_pesos = suma_pesos + peso_luna

         elif cuerpo_celeste == "marte":
              peso_marte = kg * 3.71
              print(f"el peso del cuerpo celeste elegido es: {peso_marte}")
              cant_marte = cant_marte + 1
              suma_pesos = suma_pesos + peso_marte

         elif cuerpo_celeste == "jupiter":
              peso_jupiter = kg * 24.79
              print(f"el peso del cuerpo celeste elegido es: {peso_jupiter}")
              cant_jupiter = cant_jupiter + 1
              suma_pesos = suma_pesos + peso_jupiter

    print("---RESUMEN FINAL---")
    print(f"promedio de peso final: {suma_pesos / cant_analizar}")
    print(f"cantidad de veces que se escogio la TIERRA: {cant_tierra}")
    print(f"cantidad de veces que se escogio la LUNA: {cant_luna}")
    print(f"cantidad de veces que se escogio MARTE: {cant_marte}")
    print(f"cantidad de veces que se escogio JUPITER: {cant_jupiter}")
              
bolsa()
```
