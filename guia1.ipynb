### Script 16.
Desarrolla un script que pida el peso en kilogramos y la estatura en metros de una persona. Luego,
debe calcular el Índice de Masa Corporal (IMC) usando la fórmula:

IMC = peso / (estatura²)
Después, debe mostrar la clasificación correspondiente según la OMS:
• Bajo peso: menor a 18.5
• Normal: desde 18.5 hasta 24.9
• Sobrepeso: 25.0 o más

```python
while True:
    try:
        peso = float(input("ingrese su peso en kg"))
        altura = float(input("ingrese su altura en metros"))
        break
    except ValueError:
        print("asegurese de que el peso y la altura sean numeros para continuar")

while True:
    if peso <= 0 or altura <= 0:
          print("asegurese de que el peso y la altura sean numeros positivos mayores que 0 para continuar")
          try:
                peso = float(input("ingrese su peso en kg"))
                altura = float(input("ingrese su altura en metros"))
          except ValueError:
                print("asegurese de que el peso y la altura sean numeros para continuar")
    else:
          break
    
imc = peso / (altura ** 2)
print(f"su indice de masa corporal es de {imc:.2f}")

if imc < 18.5:
    print("usted esta bajo peso")
elif imc >= 18.5 and imc <= 24.9:
    print("usted tiene un peso normal")
else:
    print("usted tiene sobrepeso")
```

### Script 17.
Desarrolla un script que pida dos números al usuario y luego solicite una operación matemática entre
las siguientes opciones: +, -, *, /.

En caso de división, debe evitar dividir por cero

```python
while True:
    try:
        num1 = float(input("ingrese un numero entero"))
        num2 = float(input("ingrese otro numero entero"))
        break
    except ValueError:
        print("asegurese de que este ingresando numeros para continuar")


while True:
    if num1 == 0 or num2 == 0:
        print("asegurese de que ambos numeros sean diferentes de 0 para continuar")
        try:
            num1 = float(input("ingrese un numero entero"))
            num2 = float(input("ingrese otro numero entero"))
        except ValueError:
            print("asegurese de que este ingresando numeros para continuar")
    else:
        break


operacion = input("ingrese la operacion a realizar(+, -, *, /): ")
while True:
    if operacion not in ["+", "-", "*", "/"]:
        print("asegurese de haber escrito +, -, * o / para continuar")
        try:
            operacion = input("ingrese la operacion a realizar(+, -, *, /): ")
        except ValueError:
            print("asegurese de haber escrito +, -, * o / para continuar")
    else:
        break

if operacion == "+":
    print(f"el resultado de la suma es {num1 + num2}")
elif operacion == "-":  
    print(f"el resultado de la resta es {num1 - num2}") 
elif operacion == "*":
    print(f"el resultado de la multiplicacion es {num1 * num2}")
else:
    print(f"el resultado de la division es {num1 / num2}")
```

### Script 20.
Desarrolla un script que calcule el peso de un objeto en distintos cuerpos celestes.
El programa debe pedir la masa del objeto en kilogramos y luego pedir el cuerpo celeste donde se
desea calcular el peso: Tierra, Luna, Marte o Júpiter.
El cálculo debe realizarse con la fórmula:

peso = masa × gravedad

Usa los siguientes valores de gravedad:

• Tierra: 9.8 m/s²

• Luna: 1.62 m/s²

• Marte: 3.71 m/s²

• Júpiter: 24.79 m/s²

El script debe mostrar la masa ingresada, el cuerpo celeste seleccionado y el peso calculado. Si el
cuerpo celeste no corresponde a una opción válida, debe mostrar: “Cuerpo celeste no válido”.

```python
while True:
    try:
        masa_cuerpo_celeste = float(input("ingrese la masa de un cuerpo celeste en kg"))
        break
    except ValueError:
        print("entrada invalida, asegurese de escribir un numero y no otra cosa")   



while True:
    try:
        cuerpo_celeste = input("ingrese el tipo de cuerpo celeste (tierra, luna, marte o jupiter): ")
        break
    except ValueError:
        print("Cuerpo celeste no válido")


while True:
    if cuerpo_celeste not in ["tierra", "luna", "marte", "jupiter"]:
        print("Cuerpo celeste no válido")
        try:
            cuerpo_celeste = input("ingrese el tipo de cuerpo celeste (tierra, luna, marte o jupiter): ")
        except ValueError:
            print("Cuerpo celeste no válido")
    else:
        break

gravedad_tierra = 9.8
gravedad_luna = 1.62
gravedad_marte = 3.71
gravedad_jupiter = 24.79

print(f"masa ingresada: {masa_cuerpo_celeste} kg")
print(f"cuerpo celeste ingresado: {cuerpo_celeste}")

if cuerpo_celeste == "tierra":
    peso = masa_cuerpo_celeste * gravedad_tierra
    print(f"el peso del cuerpo celeste en la tierra es de {peso:.2f} N")
elif cuerpo_celeste == "luna":
    peso = masa_cuerpo_celeste * gravedad_luna
    print(f"el peso del cuerpo celeste en la luna es de {peso:.2f} N")
elif cuerpo_celeste == "marte":
    peso = masa_cuerpo_celeste * gravedad_marte
    print(f"el peso del cuerpo celeste en marte es de {peso:.2f} N")
elif cuerpo_celeste == "jupiter":
    peso = masa_cuerpo_celeste * gravedad_jupiter
    print(f"el peso del cuerpo celeste en jupiter es de {peso:.2f} N")
```
