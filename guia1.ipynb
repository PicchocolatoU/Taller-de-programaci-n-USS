### Script 1

Desarrolla un script que pida un número al usuario e identifique si el valor ingresado es positivo, negativo o cero.

```python

while True:
    try:
        num = float (input("ingrese un numero"))
        break
    except ValueError:
        print("ingrese un numero no otra cosa")

if num > 0:
    print("el numero es positivo")
elif num < 0:
    print ("el num es negativo")
else:
    print ("el numero es cero")
```

### Script 2

Desarrolla un script que pida un número entero al usuario e identifique si es par o impar utilizando el operador módulo(%)


```python
while True:
    try:
        num = int(input("ingrese un numero entero"))
        break
    except ValueError:
        print("ingrese un numero entero")

num = int(input("ingrese un numero entero"))
if num % 2 == 0:
    print("el numero es par")
else:   
    print("el numero es impar")
```

### Script 6.

Desarrolla un script que pida una temperatura en grados Celsius y la clasifique como bajo cero, fría, templada o caliente. El estudiante debe definir explícitamente los rangos para cada categoría antes de programar la solución.

```python
''''
Identificador de temperatura en grados Celsius.

La temperatura se clasificará de la siguiente manera:
- Bajo cero: Menos de 0°C
- Frío: 0°C a 15°C
- Templado: 15°C a 30°C
- Caliente: Más de 30°C
'''
t = float(input"ingrese la temperatura en grados celsius")
if t < 0:
    print ("la temperatura es bajo cero")
elif t >= 0 and t <= 15:
    print ("la temperatura es fria")
elif t > 15 and t <= 30:
    print ("la temperatura es templada")
else:
    print ("la temperatura es caliente")
```

### Script 11.

Desarrolla un script para una compañía de arriendo de autos que cobra una tarifa fija de $50.000 por día.

Debes pedir al usuario la cantidad de días de arriendo y calcular el valor final según estas reglas:
- Si arrienda más de 7 días, obtiene un 25% de descuento sobre el total.
- Si arrienda más de 3 días y hasta 7 días, obtiene un 10% de descuento sobre el total.
- Si arrienda 3 días o menos, no recibe descuento.

El script debe mostrar el total a pagar

```python
a = int(input("ingrese los dias de arriendo"))
precio = a * 50000
if a > 7:
    d1 = int(precio * 0.25)
    print (f"al arrendar mas de 7 dias, se le aplicara un 25% de descuento sobre los {precio}")
    print (f"el nuevo precio a pagar es de {precio - d1}")
elif a > 3 and a <= 7:
    d2 = int(precio * 0.10)
    print(f"al arrendar entre 3 y 7 dias, se le aplicara un descuento del 10% sobre los {precio}")
    print(f"el nuevo precio a pagar es de {precio - d2}")
elif a >0 and a < 3:
    print(f"no se le aplica descuento, el precio a pagar es de{precio}")
```
